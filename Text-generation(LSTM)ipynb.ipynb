```python

import tensorflow as tf
import numpy as np

# 1. Define corpus
text = """
the sun rises in the east.
the moon shines at night.
the sky is blue.
nature is beautiful.
machine learning is interesting.
deep learning is a part of artificial intelligence.
recurrent neural networks learn patterns from sequences.
"""

text = text.lower()

# 2. Create character vocabulary
chars = sorted(set(text))
char_to_id = {c: i for i, c in enumerate(chars)}
id_to_char = {i: c for i, c in enumerate(chars)}

# 3. Convert text into numbers
data = np.array([char_to_id[c] for c in text])

# 4. Create input sequences and target characters
seq_len = 40
X, y = [], []

for i in range(len(data) - seq_len):
    X.append(data[i:i + seq_len])
    y.append(data[i + seq_len])

X = np.array(X)
y = np.array(y)

# 5. Build LSTM model
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(seq_len,)),
    tf.keras.layers.Embedding(len(chars), 32),
    tf.keras.layers.LSTM(64),
    tf.keras.layers.Dense(len(chars), activation='softmax')
])

# 6. Compile model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 7. Train model
model.fit(X, y, epochs=550, batch_size=32)

# 8. Generate text
seed = "the "
input_seq = [char_to_id[c] for c in seed]

for _ in range(100):
    pred = model.predict(
        np.array([input_seq[-seq_len:]]),
        verbose=0
    )

    next_char = id_to_char[np.argmax(pred[0])]
    seed += next_char
    input_seq.append(char_to_id[next_char])

# 9. Display generated text
print("\nGenerated Text:")
print(seed)
```
